## 2. The question: decision, action, cost of a wrong call

### Search Question
Can machine learning help prioritize which website pages should be refreshed before they lose significant search performance?

### Unit of Analysis
A single web page.

### Output
A ranked prediction score representing how strongly a page should be considered for refresh.

### Decision
Which pages should be reviewed first by the content team.

### Action
Content editors refresh the highest-ranked pages before lower-priority pages.

### Cost of a Wrong Recommendation
False positives waste editor time by refreshing healthy pages. False negatives allow genuinely declining pages to continue losing traffic, which may reduce visibility and business opportunities.

### Why ML Helps
Many signals influence search performance simultaneously, including impressions, click-through rate, content age, word count, engagement, and historical trends. Machine learning can combine these signals more effectively than a simple hand-written rule, helping prioritize limited resources.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nDeclining pages:")
print(df["trend_direction"].value_counts())

corr = df["search_volume"].corr(df["impressions_90d"])

print("\nSearch Volume vs Impressions correlation:")
print(round(corr,3))
```

```python
ctr = (
    df[df["impressions_90d"] >= 100]
    .groupby("position_tier")["ctr"]
    .mean()
)

print(ctr)
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 pages** across **44 features**.

During exploratory analysis I observed that:

- Search volume and actual impressions have almost no linear relationship (correlation ≈ **0.001**).
- Pages appearing on the first page of search results have substantially higher click-through rates than deeper pages.
- Median word count is very similar between growing and declining pages, suggesting content length alone is not the main driver of performance.

These observations suggest that deciding which pages require refresh is a multi-factor problem rather than one based on a single metric.
